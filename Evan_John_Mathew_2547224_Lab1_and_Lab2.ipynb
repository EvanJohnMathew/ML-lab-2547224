```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8,5))
sns.histplot(city['AQI'], bins=50, kde=True)
plt.title('Distribution of AQI Across Indian Cities')
plt.xlabel('AQI Value')
plt.ylabel('Frequency')
plt.show()
print("")
plt.figure(figsize=(8,2))
sns.boxplot(x=city['AQI'])
plt.title('Spread and Outliers of AQI')
plt.xlabel('AQI Value')
plt.show()

print("Mean AQI:", round(city['AQI'].mean(), 2))
print("Median AQI:", round(city['AQI'].median(), 2))
print("Skewness:", round(city['AQI'].skew(), 2))
```

**Task-5**

```python
Q1 = city['AQI'].quantile(0.25)
Q3 = city['AQI'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = city[(city['AQI'] < lower_bound) | (city['AQI'] > upper_bound)]

print("Number of Outliers:", len(outliers))
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8,3))
sns.boxplot(x=city['AQI'])
plt.title("AQI Before Outlier Treatment")
plt.show()
```

```python
city_clean = city.copy()

city_clean['AQI'] = city_clean['AQI'].clip(
    lower=lower_bound,
    upper=upper_bound
)
```

```python
plt.figure(figsize=(8,3))
sns.boxplot(x=city_clean['AQI'])
plt.title("AQI After Outlier Treatment")
plt.show()
```

```python
print("Mean AQI Before :", round(city['AQI'].mean(), 2))
print("Mean AQI After  :", round(city_clean['AQI'].mean(), 2))

print("Maximum AQI Before :", city['AQI'].max())
print("Maximum AQI After  :", city_clean['AQI'].max())
```

# **Lab-2: Data exploration and inferences**

**Task-6**

```python
import pandas as pd
import matplotlib.pyplot as plt

city['Date'] = pd.to_datetime(city['Date'])

city['Year'] = city['Date'].dt.year

yearly_aqi = city.groupby('Year')['AQI'].mean()

most_year = yearly_aqi.idxmax()
least_year = yearly_aqi.idxmin()

plt.figure(figsize=(8,5))
plt.plot(yearly_aqi.index, yearly_aqi.values, marker='o')

plt.scatter(most_year, yearly_aqi[most_year], s=100)
plt.scatter(least_year, yearly_aqi[least_year], s=100)

plt.title("Average AQI by Year")
plt.xlabel("Year")
plt.ylabel("Average AQI")
plt.grid(True)

plt.show()

print("Most Polluted Year:", most_year)
print("Least Polluted Year:", least_year)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

city['Date'] = pd.to_datetime(city['Date'])

city['Month'] = city['Date'].dt.month

monthly_aqi = city.groupby('Month')['AQI'].mean()

monthly_aqi.plot(kind='bar')

plt.title("Average AQI by Month")
plt.xlabel("Month")
plt.ylabel("Average AQI")
plt.show()

print(monthly_aqi)
```

**Task-8**

```python
aqi_state = city.groupby('State_Name')['AQI'].mean().reset_index()

crop_state = crop.groupby('State_Name')[['Area', 'Production']].mean().reset_index()

merged = pd.merge(crop_state, aqi_state, on='State_Name')

corr = merged.corr(numeric_only=True)

sns.heatmap(corr, annot=True)
plt.title("Correlation Matrix")
plt.show()

print(corr)
```
