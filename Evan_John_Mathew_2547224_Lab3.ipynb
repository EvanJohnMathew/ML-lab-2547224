```python
class LR:

    def __init__(self):
        self.m = None
        self.b = None

    def fit(self, X_train, y_train):

        num = 0
        den = 0

        for i in range(X_train.shape[0]):

            num = num + (
                (X_train[i] - X_train.mean()) *(y_train[i] - y_train.mean())
            )

            den = den + (
                (X_train[i] - X_train.mean()) * (X_train[i] - X_train.mean())
            )

        self.m = num / den

        self.b = y_train.mean() - (self.m * X_train.mean())

        print("Slope (m) =", self.m)
        print("Intercept (b) =", self.b)

    def predict(self, X_test):

        print("Input Value:", X_test)

        return self.m * X_test + self.b
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error
import numpy as np
import pandas as pd

# Re-run sklearn on CIA % using same data
X_cia = reg_df[['CIA_pct']]
y_cia = reg_df['GPA']

X_tr, X_te, y_tr, y_te = train_test_split(
    X_cia, y_cia, test_size=0.2, random_state=2   # same seed as your OLS split
)

sk_model = LinearRegression()
sk_model.fit(X_tr, y_tr)

sk_slope     = sk_model.coef_[0]
sk_intercept = sk_model.intercept_
sk_preds     = sk_model.predict(X_te)

# Manual OLS predictions on same test set
ols_slope     = lr.m          # already computed by your LR class
ols_intercept = lr.b
ols_preds     = ols_slope * X_te['CIA_pct'].values + ols_intercept

print("=" * 45)
print("       PARAMETER COMPARISON TABLE")
print("=" * 45)
print(f"{'Parameter':<15} {'Sklearn':>12} {'Manual OLS':>12}")
print("-" * 45)
print(f"{'Slope (m)':<15} {sk_slope:>12.6f} {ols_slope:>12.6f}")
print(f"{'Intercept (b)':<15} {sk_intercept:>12.6f} {ols_intercept:>12.6f}")
print("=" * 45)
```

```python
comparison_df = pd.DataFrame({
    'CIA_pct'         : X_te['CIA_pct'].values,
    'Actual_GPA'      : y_te.values,
    'Sklearn_Pred'    : sk_preds,
    'ManualOLS_Pred'  : ols_preds,
    'Difference'      : sk_preds - ols_preds
})

print(comparison_df.to_string(index=False))
print(f"\nMax difference between methods: {abs(comparison_df['Difference']).max():.8f}")
```

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Sklearn
axes[0].scatter(X_cia, y_cia, color='steelblue', label='Data points', alpha=0.7)
axes[0].plot(X_cia, sk_model.predict(X_cia), color='red', linewidth=2, label='Regression line')
axes[0].set_xlabel('CIA %')
axes[0].set_ylabel('GPA')
axes[0].set_title('Scikit-learn: CIA % vs GPA')
axes[0].legend()

# Manual OLS
axes[1].scatter(X_cia, y_cia, color='orange', label='Data points', alpha=0.7)
y_ols_line = ols_slope * X_cia['CIA_pct'].values + ols_intercept
axes[1].plot(X_cia, y_ols_line, color='red', linewidth=2, label='OLS line')
axes[1].set_xlabel('CIA %')
axes[1].set_ylabel('GPA')
axes[1].set_title('Manual OLS: CIA % vs GPA')
axes[1].legend()

plt.suptitle('Comparison: Scikit-learn vs Manual OLS', fontsize=13, fontweight='bold')
plt.tight_layout()
plt.show()
```
